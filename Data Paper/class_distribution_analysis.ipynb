```python
from google.colab import drive 
drive.mount('/content/drive')

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

CSV_PATH = "/content/drive/MyDrive/dp/data/my/csv/final_xray_metadata.csv"

df = pd.read_csv(CSV_PATH)
print(df)

class_counts = df['Class'].value_counts().sort_values(ascending=True)
class_percentages = (class_counts / class_counts.sum()) * 100

label_map = {
    "Normal": "N",
    "Pneumonia Viral": "PV",
    "Pneumonia Bacterial": "PB"
}

class_colors = {
    "Normal": "#9BC795",
    "Pneumonia Viral": "#2E5EA8",
    "Pneumonia Bacterial": "#808000"
}

labels = [label_map[c] for c in class_counts.index]
colors = [class_colors[c] for c in class_counts.index]

plt.rcParams.update({
    "font.family": "sans-serif",
    "font.size": 11
})

fig, (ax1, ax2) = plt.subplots(
    1, 2,
    figsize=(11.2, 4.6),
    gridspec_kw={"width_ratios": [1.15, 1]}
)

spacing = 0.36
bar_width = 0.14
x = np.arange(len(class_counts)) * spacing

ax1.bar(
    x,
    class_counts.values,
    width=bar_width,
    color=colors,
    edgecolor="none"
)

ax1.set_xlabel("Class", fontsize=12)
ax1.set_ylabel("Number of Images", fontsize=12)

ax1.set_xticks(x)
ax1.set_xticklabels(labels, fontsize=11)
ax1.tick_params(axis='y', labelsize=10)

for spine in ["top", "right", "left", "bottom"]:
    ax1.spines[spine].set_visible(False)

ax1.grid(axis="y", linestyle="--", linewidth=0.6, alpha=0.28)
ax1.set_ylim(0, class_counts.max() * 1.2)

ax1.text(
    0.5, -0.22, "(a)",
    transform=ax1.transAxes,
    ha="center",
    va="center",
    fontsize=13,
    fontweight="bold"
)

ax2.pie(
    class_percentages.values,
    labels=labels,
    autopct="%.1f%%",
    startangle=135,
    colors=colors,
    wedgeprops=dict(edgecolor="white", linewidth=1),
    textprops={"fontsize": 11}
)

ax2.axis("equal")

ax2.text(
    0.5, -0.22, "(b)",
    transform=ax2.transAxes,
    ha="center",
    va="center",
    fontsize=13,
    fontweight="bold"
)

plt.subplots_adjust(wspace=0.28, bottom=0.25)

plt.savefig(
    "class_distribution_bar_pie_bottom_panels.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
